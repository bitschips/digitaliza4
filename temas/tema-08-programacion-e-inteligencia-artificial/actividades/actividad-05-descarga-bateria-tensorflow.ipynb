Una entrada: tiempo en horas. Una salida: tensión en voltios. Datos ficticios para aprender regresión lineal con TensorFlow. Ejecuta las celdas en orden.

## Paso 1. Introducir los datos
Ejecuta la siguiente celda. Consulta la explicación línea por línea en la actividad de GitHub.

```python
import numpy as np

X = np.array([0, 2, 4, 6, 8, 10], dtype=np.float32)
Y = np.array([1.40, 1.38, 1.35, 1.31, 1.25, 1.12], dtype=np.float32)

print("Tiempos (horas):", X)
print("Tensiones (voltios):", Y)
```

## Paso 2. Crear la neurona
Ejecuta la siguiente celda. Consulta la explicación línea por línea en la actividad de GitHub.

```python
import tensorflow as tf

tf.keras.utils.set_random_seed(42)

modelo = tf.keras.Sequential([
    tf.keras.Input(shape=(1,)),
    tf.keras.layers.Dense(1)
])

modelo.summary()
```

## Paso 3. Preparar y entrenar
Ejecuta la siguiente celda. Consulta la explicación línea por línea en la actividad de GitHub.

```python
X_normalizado = X / 10.0

modelo.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.05),
    loss="mse"
)

historial = modelo.fit(
    X_normalizado,
    Y,
    epochs=1000,
    verbose=0
)

print("Error final (MSE):", historial.history["loss"][-1])
```

## Paso 4. Ver la gráfica de la recta aprendida

**¡Ahora verás lo que ha aprendido tu neurona!** Después de ejecutar el entrenamiento del paso 3, pulsa el botón de ejecución de la siguiente celda (▶).

Aparecerá una gráfica con **puntos** (tensiones medidas de la batería) y una **línea** (la recta de regresión calculada por la neurona). El eje horizontal indica el tiempo en horas y el vertical la tensión en voltios.

**Importante:** ejecuta primero los pasos 1, 2 y 3. Si cambias el número de épocas, vuelve a crear y entrenar la neurona antes de ejecutar esta celda para observar la nueva recta. La recta no tiene por qué pasar por todos los puntos.


```python
import matplotlib.pyplot as plt

# Generamos tiempos intermedios para dibujar la recta después del entrenamiento
horas = np.linspace(0, 10, 101)
horas_normalizadas = (horas / 10.0).astype(np.float32)
# La neurona ya entrenada predice la tensión para cada tiempo
predicciones = modelo.predict(horas_normalizadas, verbose=0).flatten()

# Los puntos son las mediciones originales
plt.scatter(X, Y, label="Datos de entrenamiento")
# La línea es la función de regresión que ha aprendido TensorFlow
plt.plot(horas, predicciones, label="Recta aprendida")
plt.xlabel("Tiempo (horas)")
plt.ylabel("Tensión (voltios)")
plt.title("Predicción de descarga de una batería")
plt.grid(True)
plt.legend()
plt.show()  # Al ejecutar esta celda se muestra la gráfica
```

## Paso 5. Interpolar
Ejecuta la siguiente celda. Consulta la explicación línea por línea en la actividad de GitHub.

```python
tiempo_nuevo = np.array([7.0], dtype=np.float32)
tension_estimada = modelo.predict(tiempo_nuevo / 10.0, verbose=0)

print(f"Tensión estimada a las 7 horas: {tension_estimada[0, 0]:.3f} V")
```

## Paso 6. Leer la ecuación aprendida
Ejecuta la siguiente celda. Consulta la explicación línea por línea en la actividad de GitHub.

```python
pesos, sesgos = modelo.layers[0].get_weights()

w = float(pesos[0, 0]) / 10.0
b = float(sesgos[0])

print(f"V(t) = {w:.5f} · t + {b:.5f}")
```

## Paso 7. Comparar entrenamientos
Ejecuta la siguiente celda. Consulta la explicación línea por línea en la actividad de GitHub.

```python
resultados = []

for n_epocas in [100, 500, 1000]:
    tf.keras.utils.set_random_seed(42)

    prueba = tf.keras.Sequential([
        tf.keras.Input(shape=(1,)),
        tf.keras.layers.Dense(1)
    ])

    prueba.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=0.05),
        loss="mse"
    )

    historia = prueba.fit(X_normalizado, Y, epochs=n_epocas, verbose=0)
    error = historia.history["loss"][-1]
    pred_7h = prueba.predict(
        np.array([0.7], dtype=np.float32), verbose=0
    )[0, 0]

    resultados.append((n_epocas, error, pred_7h))

for epocas, error, tension in resultados:
    print(f"{epocas} épocas: MSE={error:.6f}, V(7h)={tension:.3f} V")
```
